```python
import pandas as pd

matches = pd.read_csv("/Volumes/workspace/default/worldcup_raw/WorldCupMatches.csv")
players = pd.read_csv("/Volumes/workspace/default/worldcup_raw/WorldCupPlayers.csv")
cups = pd.read_csv("/Volumes/workspace/default/worldcup_raw/WorldCups_preview.csv")

print("Matches shape:", matches.shape)
print("Players shape:", players.shape)
print("Cups shape:", cups.shape)
```

```python
# evaluate nulls
print('Matches')
print(matches.isnull().sum())
print('Players')
print(players.isnull().sum())
print('Cups')
print(cups.isnull().sum())
```

```python
# Fill in the n/a values for events and position with valid reasonings
players['Event'] = players['Event'].fillna("Nothing")
players['Position'] = players['Position'].fillna("Missing")
players
```

```python
float(matches['Attendance'].mean())
```

```python
# fill n/a for attendance with the average of total attendance
matches['Attendance']=matches['Attendance'].fillna(matches['Attendance'].mean())
matches
```

```python
#drop unecessary columns ['Win conditions', 'Referee', 'Assistant 1', 'Assistant 2']
matches.drop(['Win conditions', 'Referee', 'Assistant 1', 'Assistant 2'], axis=1, inplace=True)
matches
```

```python
# convert datetime to only the date and no time
matches['Datetime'] = pd.to_datetime(matches['Datetime']).dt.date
matches.rename(columns={'Datetime': 'Date'}, inplace=True)
matches
```

```python
# make a new column to only keep track of goals
players['Goals'] = players['Event'].str.count('G')

# column for yellows 'Y'
players['Yellows'] = players['Event'].str.count('Y')

# column for reds 'R'
players['Reds'] = players['Event'].str.count('R')

# column for pentalties 'P'
players['Penalties'] = players['Event'].str.count('P')

players
```

```python
# don't need events anymore so drop
players.drop(['Event'], inplace=True, axis=1)
players
```

```python
#convert back to csv
players.to_csv("/Volumes/workspace/default/worldcup_clean/clean_players.csv")
matches.to_csv("/Volumes/workspace/default/worldcup_clean/clean_matches.csv")
cups.to_csv("/Volumes/workspace/default/worldcup_clean/clean_cups.csv")
```
